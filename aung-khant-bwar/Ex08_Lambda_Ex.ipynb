Q4) Lambda in Data Science/Data Analysis

```python
import pandas as pd

df = pd.read_csv("../data/rentals.csv")

# Q4.1) Create a new column called "size" where each entry is 
# determined by the number of bathrooms: 'large' if the number of bathrooms is greater than 3, and 'small' otherwise.


df["size"] = df["bathrooms"].apply(lambda x: "large" if x > 3 else "small")



# -----------------------------------------------------------------#
# Q4.2) Creat a new column called "price_category" where 
# Assign each entry as the following categories based on the price:
# If the price is greater than 100, the category should be 'premium'.
# Otherwise, the category should be 'standard'.

df["price_category"] = df["price"].apply(lambda x: "premium" if float(x.split("$")[1].replace(",", "")) > 100.0 else "standard")



# -----------------------------------------------------------------#
# Do not change the code below: 
# Print the first five rows to see the data. You should have 11 columns

print(f"DataFrame has {df.shape[1]} columns")
df.head()
```

# List Comprehension exercises 

```python
l = [3, 4, 4, 1, 0, 9, 3, 5, 7, 8]

diff_l = [l[idx] - l[idx+1] for idx in range(len(l)-1)]
diff_l
```

```python
password_list = ["asdfl39", "i8383jwer", "9u2jhowdjfa8dsf", "329osdof"]

valid_passwords = [password for password in password_list if len(password) >= 8]
valid_passwords
```

```python
ll = [34, 45, 3343, 3533, 222, 48934]

greater_than_100 = [num for num in ll if num > 100]
greater_than_100
```

```python
for num in range(1, 1000):
    if len(str(num)) == 1:
        print(num)
    elif len(str(num)) == 2:
        if str(num)[0] == str(num)[1]:
            print(num)
    elif len(str(num)) == 3:
        if str(num)[0] == str(num)[2]:
            print(num)
```

```python
palindromic_nums = [num for num in range(1, 1000) if str(num) == str(num)[::-1]]

palindromic_nums
```
